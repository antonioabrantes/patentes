```python
anos = [1948, 1959, 1960, 1961, 1962, 1964, 1965, 1967, 1968, 1970, 1971, 1972, 1973, 1974, 1975, 1976, 1977, 1978, 1979, 
        1980, 1981, 1982, 1983, 1984, 1985, 1986, 1987, 1988, 1989, 1990, 1991, 1992, 1993, 1994, 1995, 1996, 1997, 1998, 1999, 
        2000, 2001, 2002, 2003, 2004, 2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024]

arquivo_xlsx = "paula2024.xlsx"

for ano in anos:
    print(f"Processando .. {ano}")
    df = pd.read_excel(arquivo_xlsx, sheet_name=str(ano))
    df.columns = df.columns.str.strip()
    coluna_alvo = 'Universidade Federal do Estado do Rio de Janeiro - UNIRIO'
    ler_valores = False
    ler_codigos_area_meio = True
    for i in range(len(df)): 
        valor = df[coluna_alvo].iloc[i]
        #print(valor)
        if ler_valores:
            if pd.notna(valor):
                if str(valor)=='CÓDIGOS ÁREA FIM':
                    ler_codigos_area_meio = False
                else:
                    saida = ''
                    if ler_codigos_area_meio:
                        print('Códigos Área Meio = True')
                        saida = f'{ano};Meio'
                    else:
                        print('Códigos Área Fim = True')
                        saida = f'{ano};Fim'

                    print('CÓDIGOS ÁREA MEIO = ' + str(valor))
                    saida = saida + ';' + str(valor)
                    
                    coluna_alvo = 'Unnamed: 1'
                    valor = df[coluna_alvo].iloc[i]
                    print('CÓDIGOS NOVOS ÁREA MEIO = ' + str(valor))
                    saida = saida + ';' + str(valor)
                    
                    coluna_alvo = 'Unnamed: 2'
                    valor = df[coluna_alvo].iloc[i]
                    print('NÚMEROS DOS PROCESSOS = ' + str(valor))
                    texto_sem_ponto_virgula = ';' + str(valor).replace(";", "")
                    saida = saida + texto_sem_ponto_virgula
                    
                    coluna_alvo = 'Unnamed: 3'
                    valor = df[coluna_alvo].iloc[i]
                    print('INTERESSADO = ' + str(valor))
                    texto_sem_ponto_virgula = ';' + str(valor).replace(";", "")
                    saida = saida + texto_sem_ponto_virgula

                    coluna_alvo = 'Unnamed: 4'
                    valor = df[coluna_alvo].iloc[i]
                    print('PROCEDÊNCIA = ' + str(valor))
                    texto_sem_ponto_virgula = ';' + str(valor).replace(";", "")
                    saida = saida + texto_sem_ponto_virgula

                    coluna_alvo = 'Unnamed: 5'
                    valor = df[coluna_alvo].iloc[i]
                    print('DESCRITORES = ' + str(valor))
                    texto_sem_ponto_virgula = ';' + str(valor).replace(";", "")
                    saida = saida + texto_sem_ponto_virgula

                    coluna_alvo = 'Unnamed: 6'
                    valor = df[coluna_alvo].iloc[i]
                    print('TIPOLOGIA = ' + str(valor))
                    texto_sem_ponto_virgula = ';' + str(valor).replace(";", "")
                    saida = saida + texto_sem_ponto_virgula

                    coluna_alvo = 'Unnamed: 7'
                    valor = df[coluna_alvo].iloc[i]
                    print('LOCALIZAÇÃO = ' + str(valor))
                    texto_sem_ponto_virgula = ';' + str(valor).replace(";", "")
                    saida = saida + texto_sem_ponto_virgula
                    
                    coluna_alvo = 'Unnamed: 8'
                    valor = df[coluna_alvo].iloc[i]
                    print('OBS = ' + str(valor))
                    texto_sem_ponto_virgula = ';' + str(valor).replace(";", "")
                    saida = saida + texto_sem_ponto_virgula

                    print()
                    coluna_alvo = 'Universidade Federal do Estado do Rio de Janeiro - UNIRIO'
                    
                    with open("paula.csv", "a", encoding="utf-8") as arquivo:
                        arquivo.write(saida + "\n")

        if valor == 'CÓDIGOS ÁREA MEIO':
            #print(f"Primeiro valor: {valor}")
            ler_valores = True
```
